```python
a=input()
sum2=0
list1=list(range(1,27))

for i in range(len(a)):
    
    if i==0:
        sum2=sum2+list1[ord(a[i])-97]+list1[ord(a[i+1])-97]
    elif i==len(a)-1:
        sum2=sum2+list1[ord(a[i])-97]+list1[ord(a[i-1])-97]
    else:
        sum2=sum2+list1[ord(a[i])-97]+list1[ord(a[i-1])-97]+list1[ord(a[i+1])-97]
print(sum2)
    

```

```python
s=input()
upper=0
lower=0
number=0
special=0

for i in s:
    if len(s)>=8:
        if i.isupper():
            upper+=1
        if i.islower():
            lower+=1
        if i.isdigit():
            number+=1 
        else:
            special+=1 
if upper>0 and lower>0 and number>0 and special>0:
    print("Strong Password")
else:
    print("Weak Password")
```
