```python
names = input().split(" ")
count = 0
for word in names:
    new_word=''
    for ch in word:
        if (ch.isalpha()):
            new_word+=ch
    if new_word[0] in 'aieouAEIOU' and len(new_word)%2==0:
        count+=1 
print(count)

```
