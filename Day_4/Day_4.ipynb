```python
s=input()
n=len(s)//2
s1=s[:n]
s2=s[len(s)-n:]
count_s1=0
count_s2=0

if s1==s1[::-1]:
    count_s1=0 
else:
    l=0
    h=len(s1)-1
    while(l<h):
        if s1[l]!=s1[h]:
            count_s1+=1 
        l+=1
        h-=1
if s2==s2[::-1]:
    count_s2=0 
else:
    l=0
    h=len(s2)-1
    while(l<h):
        if s2[l]!=s2[h]:
            count_s2+=1 
        l+=1
        h-=1

print(count_s1+count_s2)
```
